```python
# Creating a DataFrame from a dictionary of Series

import pandas as pd

info = {
    'one': pd.Series(
        [1, 2, 3, 4, 5, 6],
        index=['a', 'b', 'c', 'd', 'e', 'f']
    ),

    'two': pd.Series(
        [1, 2, 3, 4, 5, 6, 7, 8],
        index=['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h']
    )
}

d1 = pd.DataFrame(info)

print(d1)
```

```python
# Adding a new column to a DataFrame

import pandas as pd

info = {
    'one': pd.Series([1, 2, 3, 4, 5],
                     index=['a', 'b', 'c', 'd', 'e']),

    'two': pd.Series([1, 2, 3, 4, 5, 6],
                     index=['a', 'b', 'c', 'd', 'e', 'f'])
}

df = pd.DataFrame(info)

print(df)
```

```python
#column deletion
import pandas as pd

info = {'one' : pd.Series([1, 2], index= ['a', 'b']),
'two' : pd.Series([1, 2, 3], index=['a', 'b', 'c'])}

df = pd.DataFrame(info)
print ("The DataFrame:")
print (df)

# using del function
print ("Delete the first column:")
del df['one']
print (df)
# using pop function
print ("Delete the another column:")
df.pop('two')
print (df)
```

```python
# Question: Select rows from a DataFrame using row labels with loc

import pandas as pd

info = {
    'one': pd.Series([1, 2, 3, 4, 5],
                     index=['a', 'b', 'c', 'd', 'e']),
    'two': pd.Series([1, 2, 3, 4, 5, 6],
                     index=['a', 'b', 'c', 'd', 'e', 'f'])
}

df = pd.DataFrame(info)

print("Data frame:")
print(df)

print("b index elements are:")
print(df.loc['b'])

print("e index elements are:")
print(df.loc['e'])
```

```python
# Question: Select a row from a DataFrame using integer location with iloc

import pandas as pd

info = {
    'one': pd.Series([1, 2, 3, 4, 5],
                     index=['a', 'b', 'c', 'd', 'e']),
    'two': pd.Series([1, 2, 3, 4, 5, 6],
                     index=['a', 'b', 'c', 'd', 'e', 'f'])
}

df = pd.DataFrame(info)

print("Data frame:")
print(df)

print("Row at integer position 3:")
print(df.iloc[3])
```

```python
# Select multiple rows from a DataFrame using slicing

import pandas as pd

info = {
    'one': pd.Series([1, 2, 3, 4, 5],
                     index=['a', 'b', 'c', 'd', 'e']),
    'two': pd.Series([1, 2, 3, 4, 5, 6],
                     index=['a', 'b', 'c', 'd', 'e', 'f'])
}

df = pd.DataFrame(info)

print("Data frame:")
print(df)

print("Rows from position 2 to 4:")
print(df[2:5])
```

```python
# Add new rows to an existing DataFrame using concat()

import pandas as pd

# First DataFrame
d = pd.DataFrame({
    'A': [1, 2, 3],
    'B': [4, 5, 6]
})

# Second DataFrame
d2 = pd.DataFrame({
    'A': [7, 8, 9],
    'B': [10, 11, 12]
})

print("First DataFrame:")
print(d)

print("Second DataFrame:")
print(d2)

# Add rows of d2 to d
d = pd.concat([d, d2], ignore_index=True)

print("DataFrame after adding rows:")
print(d)
```

```python
# Delete rows from a DataFrame using the drop() function

import pandas as pd

a_info = pd.DataFrame(
    [[4, 5], [6, 7]],
    columns=['x', 'y']
)

b_info = pd.DataFrame(
    [[8, 9], [10, 11]],
    columns=['x', 'y']
)

# Combine both DataFrames
a_info = pd.concat([a_info, b_info])

print("DataFrame before deletion:")
print(a_info)

# Drop rows with index label 0
a_info = a_info.drop(0)

print("DataFrame after deleting row with label 0:")
print(a_info)
```
