```python
#part 15 - text analysis using nummpy
#tokenization
# Store the given sentence.
sentence2 = "Data Science and Machine Learning use NumPy arrays every single day"

# Convert the sentence to lowercase and split it into individual words.
tokens2 = sentence2.lower().split()

# Display the tokens.
print(tokens2)
```

```python
#building a vocabulary
# Convert tokens2 into a NumPy array.
tokens2_array = np.array(tokens2)

# Find all unique words in tokens2.
vocabulary2 = np.unique(tokens2_array)

# Display the vocabulary.
print("Vocabulary:", vocabulary2)

# Count the number of unique words.
print("Vocabulary size:", len(vocabulary2))
```

```python
#word frequency vector
# Convert tokens2 into a NumPy array.
tokens2_array = np.array(tokens2)

# Count how many times each vocabulary word appears.
vec2 = np.array([
    np.sum(tokens2_array == word)
    for word in vocabulary2
])

# Display the vocabulary.
print("Vocabulary:", vocabulary2)

# Display the bag-of-words vector.
print("Bag-of-words vector:", vec2)
```

```python
#compare two sentences
# Create two sentences that are similar in meaning and topic.
sentence1 = "I enjoy learning Python programming"
sentence2 = "I like studying Python coding"

# Tokenize the first sentence.
tokens1 = sentence1.lower().split()

# Tokenize the second sentence.
tokens2 = sentence2.lower().split()

# Create one shared vocabulary from both sentences.
combined_vocab = np.unique(np.array(tokens1 + tokens2))

# Create the bag-of-words vector for the first sentence.
vec1 = np.array([
    np.sum(np.array(tokens1) == word)
    for word in combined_vocab
])

# Create the bag-of-words vector for the second sentence.
vec2 = np.array([
    np.sum(np.array(tokens2) == word)
    for word in combined_vocab
])

# Calculate cosine similarity between the two vectors.
cosine_sim = np.dot(vec1, vec2) / (
    np.linalg.norm(vec1) * np.linalg.norm(vec2)
)

# Display the similarity score.
print("Cosine similarity:", cosine_sim)
#Cosine similarity is close to 1 when two vectors point in a similar direction, meaning the sentences have similar word patterns. It is close to 0 when the vectors point in different directions, meaning the sentences are unrelated. A cosine similarity of exactly 1 means the two vectors point in exactly the same direction.
```

```python
#most frequent words
# Store a paragraph containing 3 sentences.
paragraph = "Python is useful for data science. Python is popular for data analysis. Data science uses Python for analysis."

# Convert the paragraph to lowercase and remove periods.
clean_text = paragraph.lower().replace(".", "")

# Split the paragraph into individual words.
tokens = clean_text.split()

# Convert the tokens into a NumPy array.
tokens_array = np.array(tokens)

# Build the vocabulary using unique words.
vocabulary = np.unique(tokens_array)

# Count how many times each vocabulary word appears.
word_counts = np.array([
    np.sum(tokens_array == word)
    for word in vocabulary
])

# Sort the indices according to word frequency.
sorted_indices = np.argsort(word_counts)

# Reverse the order so the most frequent words come first.
sorted_indices = sorted_indices[::-1]

# Select the indices of the top 3 most frequent words.
top_3_indices = sorted_indices[:3]

# Print the top 3 words and their frequencies.
for index in top_3_indices:
    print(vocabulary[index], ":", word_counts[index])
```

```python
#AP24110011267
#Mini Project — Simple Text Similarity Checker
# Import NumPy for array and numerical operations.
import numpy as np

# Enter three short sentences.
sentence1 = "I love learning Python programming"
sentence2 = "Python programming is very interesting"
sentence3 = "The weather is beautiful today"

# Tokenize the first sentence.
tokens1 = sentence1.lower().split()

# Tokenize the second sentence.
tokens2 = sentence2.lower().split()

# Tokenize the third sentence.
tokens3 = sentence3.lower().split()

# Combine all words from the three sentences.
all_tokens = tokens1 + tokens2 + tokens3

# Build one shared vocabulary for all three sentences.
vocabulary = np.unique(np.array(all_tokens))

# Create the bag-of-words vector for sentence 1.
vector1 = np.array([
    np.sum(np.array(tokens1) == word)
    for word in vocabulary
])

# Create the bag-of-words vector for sentence 2.
vector2 = np.array([
    np.sum(np.array(tokens2) == word)
    for word in vocabulary
])

# Create the bag-of-words vector for sentence 3.
vector3 = np.array([
    np.sum(np.array(tokens3) == word)
    for word in vocabulary
])

# Calculate cosine similarity between sentence 1 and sentence 2.
similarity_12 = np.dot(vector1, vector2) / (
    np.linalg.norm(vector1) * np.linalg.norm(vector2)
)

# Calculate cosine similarity between sentence 1 and sentence 3.
similarity_13 = np.dot(vector1, vector3) / (
    np.linalg.norm(vector1) * np.linalg.norm(vector3)
)

# Calculate cosine similarity between sentence 2 and sentence 3.
similarity_23 = np.dot(vector2, vector3) / (
    np.linalg.norm(vector2) * np.linalg.norm(vector3)
)

# Display all three similarity scores.
print("Similarity between Sentence 1 and Sentence 2:", similarity_12)
print("Similarity between Sentence 1 and Sentence 3:", similarity_13)
print("Similarity between Sentence 2 and Sentence 3:", similarity_23)

# Store the three similarity scores in an array.
similarities = np.array([
    similarity_12,
    similarity_13,
    similarity_23
])

# Find the position of the highest similarity score.
most_similar_index = np.argmax(similarities)

# Identify the pair with the highest similarity.
if most_similar_index == 0:
    print("Most similar pair: Sentence 1 and Sentence 2")
elif most_similar_index == 1:
    print("Most similar pair: Sentence 1 and Sentence 3")
else:
    print("Most similar pair: Sentence 2 and Sentence 3")
```

What information does a bag-of-words vector lose about the original sentence?
